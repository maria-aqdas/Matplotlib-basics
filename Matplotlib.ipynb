```python
import matplotlib.pyplot as plt
```

```python
x=[1,2,3,4,5]
y=[10,45,76,84,34]
plt.plot(x,y)
plt.show()
```

```python
plt.figure(figsize=(3,3))
plt.plot(x,y, color="green",marker="o",linestyle="dashed", linewidth=2, markersize =15)
plt.title("My First Matplotlib Chart")
plt.xlabel("Its x-axis")
plt.ylabel("Its y-asix")
plt.show()
```

```python
x=[1,2,3,4,5]
y1=[20,30,40,50,80]
y2=[40,60,90,85,70]
plt.plot(x,y1,label="Sales of 2024")
plt.plot(x,y2,label="Sales of 2025")
plt.title("Year on Year Sale")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
          
```

```python
#Bar chart
x=['a','b','c','d','e']
y=[23,45,65,76,34]
plt.bar(x,y)
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.title("Bar Chart Example")
plt.show()
```

```python
#histogram
import random
data=[random.randint(1,20) for r in range(500)]
plt.hist(data,bins=12)
plt.title("Histogram Example")
plt.show()
```

```python
#Pie Chart
catagory=["A","B","C","D","E"]
sales=[76,56,48,98,76]
plt.pie(sales, labels=catagory, autopct="%1.1f%%")
plt.title("Pie Chart Example")
plt.show()
```

```python
#Scatter chart
a=[3,6,7,8,4]
b=[34,54,32,36,76]
plt.scatter(a,b)
plt.title("Scetter chart example")
plt.show()
```

```python
#Subplot
catagory=["A","B","C","D","E"]
sales=[76,56,48,98,76]

a=[3,6,7,8,4]
b=[34,54,32,36,76]

plt.figure(figsize=(5,5))
plt.subplot(1,2,1)
plt.bar(catagory,sales)
plt.title("bar chart")
plt.xlabel("catagory values")
plt.ylabel("sales values")

plt.subplot(1,2,2)
plt.scatter(a,b)
plt.title("scetter chart")
plt.xlabel("a values")
plt.ylabel("b values")
```
